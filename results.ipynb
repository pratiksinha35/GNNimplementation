```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
classification_df = pd.read_csv('results/classification.csv')
regression_df = pd.read_csv('results/regression.csv')
```

```python
classification_df
```

```python
regression_df
```

```python
df = regression_df.copy()

# Plot settings
fig, ax = plt.subplots(figsize=(10, 6))

# Set width of bars
bar_width = 0.35
index = np.arange(len(df))

# Create bars
bar1 = ax.bar(index, df['MPNN'], bar_width, 
              label='MPNN', color='steelblue', alpha=0.8)
bar2 = ax.bar(index + bar_width, df['MetaModel'], bar_width, 
              label='MetaModel', color='coral', alpha=0.8)

# Add labels and title
ax.set_xlabel('Dataset', fontsize=12, fontweight='bold')
ax.set_ylabel('Score', fontsize=12, fontweight='bold')
ax.set_title('Model Performance Comparison', fontsize=14, fontweight='bold')
ax.set_xticks(index + bar_width / 2)
ax.set_xticklabels(df['Dataset'])
ax.legend()

# Add value labels on top of bars
def autolabel(bars):
    """Attach a text label above each bar displaying its height."""
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height:.3f}',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=10)

autolabel(bar1)
autolabel(bar2)

# Adjust layout
plt.tight_layout()
plt.show()
```
